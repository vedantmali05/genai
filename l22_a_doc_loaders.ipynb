# Complete RAG Flow

1. Preprocessing
    1. **Document Collection:**
        - Collect required documents
        - This could be policies or financial documents in businesses, references in research writings, notes for academic studies, etc.
    2. **Document Loading:**
        - Extract text from document.
        - Different documents like PDFs, Word Files, Excel Sheets, etc. contain many graphical and structure information like images, borders, watermarks, links, etc.
        - LLMs can only read text, so we only extract the textual information from the document.
    3. **Document Chunking:**
        - Breaking the doc into smaller docs.
        - We may break a doc by a group of pages, by chapters, by single page each, by paragraphs, by sentences, etc. based on business requirement.
    4. **Embedding:**
        - Converting the chunks into **embeddings** or **vector arrays**
        - See [./l21_embeddings_and_cosine_similarity.ipynb](./l21_embeddings_and_cosine_similarity.ipynb)
    5. **Storage:**
        - Vectors once created, are stored into the Vector Stores or Vector DBs.
2. **Retrieval**
3. **Augmentation**
4. **Generation**

# Document Loaders
- Different documents are written in different formats and encodings. E.g.:
    - A **PDF file** contains structured visual layouts, fonts, and vector streams.
    - A **Word document (.docx)** is an XML-based archive (OpenXML).
    - A **plain text file (.txt)** contains raw characters in specific encodings (like UTF-8 or ASCII).
    - A **Markdown file (.md)** contains plain text formatted with Markdown syntax tags.
    - A **webpage / URL** is written in HTML with embedded CSS and JavaScript scripts.
    - A **CSV / Excel sheet** stores tabular, comma-separated, or binary cell grids.
    - A **JSON file** stores structured key-value pairs and nested objects.
- We need to extract only **unstructured plaintext** from these documents.
- We use **Document Loaders** (e.g., `PyPDFLoader`, `Docx2txtLoader`, `UnstructuredHTMLLoader` in LangChain) for this purpose.

## Available Doc Loaders in LangChain

### 1. Basic & Local Files (Easiest)
- **`TextLoader`**: Reads plain text files directly into single Document objects.
- **`DirectoryLoader`**: Scans folders and loads multiple files automatically based on file extensions.
- **`CSVLoader`**: Loads CSV files row-by-row into structured key-value text.
- **`UnstructuredMarkdownLoader`**: Parses Markdown files into text while respecting headers and lists.
- **`Docx2txtLoader`**: Extracts plain text directly from Microsoft Word documents.

### 2. PDF Loaders
- **`PyPDFLoader`**: Lightweight, pure-Python loader to read PDFs page-by-page.
- **`OpenDataLoader PDF` (`OpenDataLoaderPDFLoader`)**: Fast, modern PDF loader designed for RAG that extracts text, tables, and optional markdown directly from local files or directories.
- **`pdfmuse` (`PdfmuseLoader`)**: Deterministic parser for PDFs and DOCX files that extracts clean text, exact coordinates, tables, and section-level metadata for RAG.
- **`PyMuPDFLoader`**: High-speed, accurate PDF parser using the MuPDF engine.

### 3. Webpages & Online Data
- **`WebBaseLoader`**: Uses BeautifulSoup to scrape and extract text from public URLs.
- **`UnstructuredHTMLLoader`**: Strips HTML tags and script elements to grab clean webpage body text.
- **`YoutubeLoader`**: Fetches video transcripts directly from YouTube videos into readable text.
- **`FirecrawlLoader` / `ApifyDatasetLoader`**: Handles complex JavaScript-rendered websites and multi-page web crawls.

## OpenDataLoaderPDFLoader

```python
from langchain_opendataloader_pdf import OpenDataLoaderPDFLoader

documents = OpenDataLoaderPDFLoader(file_path="./media/recipies.pdf", format="text").load()
```
