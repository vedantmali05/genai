# Prompt Engineering

A practice of designing and refining instructions. Produce more accurate, reliable and useful outputs.

## Structure of a Good Prompt

Goal (important):
- What we wanna to?

Context (helpful):
- Background information that the AI may need for clarity

Contraints (mostly critical):
- Forced rules, guidelines and instructions to say what not to do.
- E.g.: Format, language tone, rules, etc.

Examples (optional, but helpful):
- LLMs learn how to reply by studying the examples given by the user, on the go.
- Based on how examples are given, the types are: Zero-shot prompting (no examples), One-shot prompting, few-shot prompting.

Role (optional, but helpful):
- When specified a role, the LLM starts to behave like it.

### Example

```python
from IPython.display import Markdown # Just to print AI response Beautifully

from dotenv import load_dotenv
load_dotenv()


from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
user_input = "Tea"

system_prompt = """
Role: You are a professional chef.
Goal: Write a step-by-step recipe for the requested food item.
Context: The user wants to prepare a meal quickly and needs precise steps.
Constraints:
- List the steps numerically.
- Do not include any extra conversational filler, introduction, or jargon text.
"""

messages = [
    ("system", system_prompt),
    ("human", user_input)
]

ai_message = llm.invoke(messages)
Markdown(ai_message.content)
```

## Types of Prompting

1. Zero-shot Prompting:
2. One-shot Prompting:
3. Few-shot Prompting:
4. Chain of Thoughts:
5. Tree of Thoughts:
6. Persona/Role:

### 1. Zero-shot prompt

- No example given to the prompt.

```python
user_input = "Tea"

system_prompt = "Write a step-by-step recipe for the given food item."

messages = [
    ("system", system_prompt),
    ("human", user_input)
]

ai_message = llm.invoke(messages)
Markdown(ai_message.content)
```

### 2. One-shot Prompting

- One example is given

```python
user_input = "Tea"

system_prompt = "Write a step-by-step recipe for the given food item."

messages = [
    ("system", system_prompt),
    # One-shot example (Human & AI message pair)
    ("human", "Coffee"),
    ("ai", "1. Boil water.\n2. Add instant coffee powder to a cup.\n3. Pour water.\n4. Stir and serve."),
    # Actual input
    ("human", user_input)
]

ai_message = llm.invoke(messages)
Markdown(ai_message.content)
```

### 3. Few-shot Prompting

- Different examples are given showing different cases

```python
user_input = "Tea"

system_prompt = "Write a step-by-step recipe for the given food item."

messages = [
    ("system", system_prompt),
    # Few-shot example 1 (Coffee)
    ("human", "Coffee"),
    ("ai", "1. Boil water.\n2. Add instant coffee powder to a cup.\n3. Pour water.\n4. Stir and serve."),
    # Few-shot example 2 (Lemonade)
    ("human", "Lemonade"),
    ("ai", "1. Squeeze fresh lemon juice into a glass.\n2. Add sugar and water.\n3. Stir until sugar dissolves.\n4. Add ice and serve."),
    # Actual input
    ("human", user_input)
]

ai_message = llm.invoke(messages)
Markdown(ai_message.content)
```

### 4. Chain of Thoughts (CoT)

- Explaining the step-by-step reasoning/thought process before presenting the final output.

```python
user_input = "Tea"

system_prompt = """
You are a culinary expert. When asked for a recipe:
1. Think step-by-step about the key preparation phases, timing, and why we do them in that order.
2. Outline this reasoning first under a 'Reasoning:' section.
3. Finally, present the recipe steps under a 'Recipe:' section.
"""

messages = [
    ("system", system_prompt),
    ("human", user_input)
]

ai_message = llm.invoke(messages)
Markdown(ai_message.content)
```

### 5. Tree of Thoughts (ToT)

- Exploring and evaluating multiple reasoning paths or approaches before arriving at the final conclusion.

```python
user_input = "Tea"

system_prompt = """
You are an expert chef. Explore three different preparation methods or paths for brewing the requested drink.
For each method:
1. Outline the key steps briefly.
2. Evaluate its pros and cons (e.g., taste intensity, effort, speed).
Compare all three methods and recommend the single best one for a quick breakfast.
"""

messages = [
    ("system", system_prompt),
    ("human", user_input)
]

ai_message = llm.invoke(messages)
Markdown(ai_message.content)
```

### 6. Persona/Role

- Instructing the LLM to adopt a specific character, role, or tone.

```python
user_input = "Tea"

system_prompt = """
Role: You are a traditional British Royal Butler.
Goal: Write a step-by-step recipe for the requested food or drink item.
Tone: Extremely polite, formal, and matching royal tea-serving etiquette.
"""

messages = [
    ("system", system_prompt),
    ("human", user_input)
]

ai_message = llm.invoke(messages)
Markdown(ai_message.content)
```
