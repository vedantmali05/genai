# Retrieval Augmented Generation (RAG)

- AI Models are trained on **past data**, typically, 6 months older than it's deployment/release date.
- E.g., GPT Luna model is trained till Feb 16, 2026 only.
- This date is known as **Knowledge Cutoff Date**
- This problem of knowledge cutoff can be solved by giving the model access to internet, where, it can make searches and retrieve context
- But, in case where the data is not available on the internet, the AI will be no longer have knowledge about it.
- For e.g., a company's policy docs, governance docs, terms and conditions, workflows, project plan, etc. are private and never shared publicly on the internet.
- Such docs are to be provided to the AI separately.
- This process of providing **external** or **unknown** knowledge to the AI, is known as, RAG.

## Retrieval in RAG

- Imagine if we provide 1000 pages book about India to AI, and asked a question *"How many states are there in India?"*
- Then, the AI model would need to read the entire 1000 page document to find the answer of the given question.
- Thus, context-window size will exceed, token limits will reach, and processing time will be very high for just a simple prompt.

**Retrieval is a solution to this.**
1. Divide the document into text chunks
    ```plaintext
    [
        ch1
        ch2
        ch3
        ...
    ]
    ```
1. Convert every text chunk $\rightarrow$ Vector Embeddings
    ```plaintext
    [
        ch1 -> v1
        ch2 -> v2
        ch3 -> v3
        ...
    ]
    ```
1. Convert User Prompt $\rightarrow$ Vector Embeddings
    ```plaintext
    vp
    ```
1. Compare every chunk $\leftrightarrow$ user prompt
1. While comparing, calculate: **Cosine Similarity**
    ```plaintext
    [
        v1 x vp = some cosine similarity
        v2 x vp = some cosine similarity
        v3 x vp = some cosine similarity
        ...,   ...     ...
    ]
    ```
1. The text chunk with maximum cosine similarity with the user prompt is selected.

Thus, to the AI model, we only provide 2 things:
1. User prompt, of course.
2. Selected text chunk

.... instead of sharing the 1000 page document

**NOTE: This process of finding the most similar part of a document that matches with the input user query or prompt is known as Semantic Search**

### Difference between Retrieval and Semantic Search
- Retrieval is a big process and has many sub-processes in it.
- Semantic search is a small sub-process inside the process of Retrieval
