```python
import random
```

```python
class Car:
    # Class attr -
    # Same value is available accross all the 
    offers_available = True
    
    # Constructor
    def __init__(self, brand: str, color: str, milageKmps: float):
        # Private attr, starts with __
        # Protected attr, starts with _
        # The protected attribute works exactly like the public ones, but, acts as a convention/signal for the devs to not to use it outside the class.
        
        self.__card_id = random.randint(1000, 9999)
        self._brand = brand
        self.color = color
        self.milage = milageKmps
    
    def features(self):
        print(f"{'ID:':>6} {self.__card_id}\n{'Brand:':>6} {self._brand}\n{'Color:':>6} {self.color}\n{'Milage:':>6} {self.milage}")
        if (Car.offers_available):
                print(f"{'*Offers Available':>6}")
    
    def change_color(self, color: str):
        print(f"Color changed from {self.color} to ", end="")
        self.color = color
        print(f"{self.color}.")
```

```python
car1 = Car("Jaguar", "White", 30)
```

```python
car1.features()
car1.change_color("Black")
# car1.__card_id <- will throw error, isn't accessible directly
```
