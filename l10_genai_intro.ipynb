# Introduction to GenAI

## What's a parameter (or weight)?
- A parameter (or weight $w_i$) in an LLMs is just a decimal number (floating-point number)
- Example:
```python
    [
        # Delhi is capital of India
        [ 0.0012, -0.4521,  0.8912, -0.0124],
        # The city sits on the bank of the Yamuna River
        [-0.1254,  0.3129, -0.0051,  0.6120],
    ]

    # When a new parameter is added during training:
    [
        # Delhi is capital of India
        [ 0.0012, -0.4521,  0.8912, -0.0124],
        # The city sits on the bank of the Yamuna River
        [-0.1254,  0.3129, -0.0051,  0.6120],
        # The population of Delhi is estimated at approximately 2.33 crore
        [ 0.7841, -0.9122,  0.1143, -0.3341]
    ]
```

## AI Model

An AI model contains
- billions or even trillions of parameters (or weights $w_i$)
- takes inputs ($x_i$)
- to generate output ($y$),

which can be represented by the following equation:
$$
y = w_1x_1 + w_2x_2 + \dots + w_ix_i \\ \text{(These can be trillions)}
$$

For example, let
- The inputs be $x_1 = 2$, $x_2 = 10$, and $x_3 = 2$, 
- Their corresponding weights (parameters) be $w_1 = 3$, $w_2 = 0.5$, and $w_3 = 4$. 
- Then, the output $y$ is computed as:
$$
y = (3 \times 2) + (0.5 \times 10) + (4 \times 2) \\ = 6 + 5 + 8 \\ = 19
$$

In real-world GenAI models, this calculation is repeated trillions of times across multiple layers to generate complex outputs like text, images, or code.

## Supervised v/s Unsupervised Learning

- **Supervised Learning**: For training, we have both, the problem data and actual outputs.

| Problems | Actual Outputs | Predicted Outputs | Is Correct |
|---|---|---|---|
| ABC | 42 | 42 | ✅ |
| DEF | 88 | 75 | ❌ |
| GHI | 15 | 15 | ✅ |

- **Unsupervised Learning**: For training, we only have, the problem data.

| Problems | Predicted Outputs | Is Correct |
|---|---|---|
| ABC | 42 | ❓ (Model decides itself) |
| DEF | 75 | ❓ (Model decides itself) |
| GHI | 19 | ❓ (Model decides itself) |

## Fine Tuning

Fine-tuning means taking an already trained general AI model and training it for a specific task.

- **Example**: Taking a general LLM (like Gemini or GPT) and training it for only 1 task, Automatic Flight Recommendation and Booking.
