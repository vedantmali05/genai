# `tiktoken` Quick Guide

**`tiktoken`** is OpenAI's fast tokenizer. `tiktoken` converts given text to tokens.

### Key Concepts
- **Encoding**: Text $\rightarrow$ List of Token IDs.
- **Decoding**: List of Token IDs $\rightarrow$ Text.


```python
import tiktoken

# 1. Load the tokenizer for a specific model
encoding = tiktoken.encoding_for_model("gpt-4o")
```

```python
text = "Meow Meow"

# 2. Encode: Text to Token IDs
tokens = encoding.encode(text)

print("Original Text:", text)
print("Token IDs:", tokens)
print("Token Count:", len(tokens))
```

# Types of Messages in GenAI

Inputs
- **Human Message (aka. User - Mandatory)**: The prompt, question, or request typed by the user.
- **System Message (Optional)**: Background instructions given to set the AI's behavior, role, or rules.

Output
- **AI Message (aka. Assistant)**: The answer or response created by the AI for the user.

```python
from IPython.display import Markdown # Just to print AI response Beautifully

from dotenv import load_dotenv
load_dotenv()

from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

## Type 1: Simple String

```python
# Human Message (Prompt given by user)
user_input = input("Enter recipe name: ")

# System Message (Extra instructions given by System)
system_message = f"""
    Write the recipe for {user_input} in step-by-step fashion.
    Don't give any extra jargon text.
"""

# AI Message (Response)
ai_message = llm.invoke(system_message)
Markdown(ai_message.content)
```

## Type 2: message Object (`langchain_core.messages`)
- The previous approach to implement Human, System and AI messages is not preffered.
- We should use modern techniques like `HumanMessage()`, `SystemMessage()` and `AIMessage()`, `dict_template`, etc.

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

user_input = input("Enter recipe name: ")

# A list of Human, System and AI Messages
messages = [
    # System Message
    SystemMessage("Write the recipe for given food item in step-by-step fashion. Don't give any extra jargon text."),
    
    # Giving an example to AI: Tea Recipe
    # Human Message
    HumanMessage("Tea"),
    # AI Message
    AIMessage("1. Boil water.\n2. Steep tea bag.\n3. Serve."),

    # Actual Human Message
    HumanMessage(user_input)
]

ai_message = llm.invoke(messages)
Markdown(ai_message.content)
```

# Type 3: Dictionary (OpenAI chat format)

A list of `{"role": ..., "content": ...}` dicts is provided.

Roles: `user`, `system`, `assistant`.

```python
user_input = input("Enter recipe name: ")

# A list of Human, System and AI Messages
messages = [
    # System Message
    {"role": "system", "content": "Write the recipe for given food item in step-by-step fashion. Don't give any extra jargon text."},
    
    # Giving an example to AI: Tea Recipe
    # Human Message
    {"role": "user", "content": "Tea"},
    # AI Message
    {"role": "assistant", "content": "1. Boil water.\n2. Steep tea bag.\n3. Serve."},

    # Actual Human Message
    {"role": "user", "content": user_input},
]

assistant_message = llm.invoke(messages)
Markdown(assistant_message.content)
```

## Type 4: Tuples

A list of `{role: ..., content: ...}` tuples is provided.

Roles: `human`, `system`, `ai`.

```python
user_input = input("Enter recipe name: ")

# A list of Human, System and AI Messages
messages = [
    # System Message
    ("system", "Write the recipe for given food item in step-by-step fashion. Don't give any extra jargon text."),
    
    # Giving an example to AI: Tea Recipe
    # Human Message
    ("human", "Tea"),
    # AI Message
    ("ai", "1. Boil water.\n2. Steep tea bag.\n3. Serve."),

    # Actual Human Message
    ("human", user_input),
]

ai_message = llm.invoke(messages)
Markdown(ai_message.content)
```
