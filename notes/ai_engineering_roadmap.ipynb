Here it starts...

## 1. Large Language Models:

The central "brain" of any AI system. We must:

- Understand how they generate text which is: Unpredictable and Non-Deterministic
- Which model fits the particular task?

Every model is trained on a different kind of data, and optimized for a different kind of output.

Know what the model can do, what it costs, and how fast it runs.


## 2. Wrapping Layer - Making It Useful:

A LLM can ingest queries and return responses, but they don't understand our ecosystem.

To turn an LLM into an AI agent that performs real work, you must wrap it in three essential components:

### a. Tools:

Allow the model to interact with external systems like databases, APIs or programs via protocols like 

MCP (Model Context Protocol) lets any LLM to connect any tool.

### b. External Knowledge (RAG):

Uses Retrieval-Augmented Generation (RAG) to give the LLM access to specific, private, or current information which was invisible during it's training.

RAG is built on two concepts:
- Retrieval: Pull relevant information from external knowledge base
- Augmented Generation: Use that information to generate relevant responses.

Every RAG model is pretty much an improvement in one of the two halves.

### c. Memory:

LLM has no memory across sessions. Memory = pretty much how you fix that.

This is categorized into:
- Short-term memory: Conversation history within a single session.
- Long-term memory: Data that persists across sessions, further split into:
    - Semantic Memory: facts about a user
    - Episodic Memory: specific events or episodes related to the user.


Thus, an AI Agent has been created now 🤓, where we have answered:

- What does it needs?
- What does it needs to know?
- What does it needs to remember?

But, we usually forget an important question: **How do you know any of it is actually working?**

## 3. Evals - Proving It Works:

- We build something, it feels fine, we ship it, and later something breaks, which then costs more...
- Thus, use Evals - Catch that even before it happens. This a single skill that can keep the product working in the market.
- E.g., Air Canada's AI Chatbot messed up by confidently showing wrong information that contradicts with their own policy, which required the feature to be taken down later.

### Working of Evals:

Testing in AI is very complicated as LLMs are unpredictable, and there are many right answers and many wrong, unlike the traditional programs.

Evals is a way of defining quality and measuring the performance. 

LLM-as-a-judge (AI Judges) is an implementation of Evals where we give sample queries and "Rubrics" of responses. Rubric = set of rules, guidelines, or criterias. Other AI agents are set up as judges which iteratively validated the responses against each rubric, and rate the performance.

Often, AI Agents also have curated dataset or samples of examples which support this process. Faster than human-review.

## 4. Multi-Agent Systems - Scaling Up:

- Every AI agent hits limit, solution: add a new agent on top of it.
- Do not over-engineer by adding agents too early.
- Use a multi-agent architecture only when a task requires:
Specialization (distinct skills)
Parallelization (handling multiple sub-tasks simultaneously)
and more...

## 5. AI Ops - Life After Launch:
- AI systems are never truly "finished."
- The AI Ops cycle involves constant design, build, evaluate, deploy, and monitor loops.
- Systems improve over time based on real-world usage signals.

## 6. Thinking:
- System Design = a fancy word for thinking.
- A human need to THINK, before a AI does it wrong.
- THINKING is the most important Human Skill
- How to think System Design:
    1. Walk through the map
    2. Pick the parts that fit the problem
    3. Make trade-offs at each step
