# Multimodal LLMs

The flow of the code will be like this:

1. Imports and Initialisation
1. Get Image **Path** or **URL**
1. Convert Image &rarr; **Image Bytes**
1. Convert Image Bytes &rarr; **Base64 Bytes**
1. Convert Base64 Bytes to &rarr; **Base64 String** (utf-8) format
1. Provide Base64 String to LLM. We also need to specify **MIME Type**.
1. Get output and enjoy

Yo, the Bytecode, Base64, MIME Type, utf-8, etc. everything is explained below, don't hurry, just read.

## Step 1: Imports and Initialisation

```python
# For getting the path
from pathlib import Path
# For converting to Base64
import base64

from dotenv import load_dotenv
load_dotenv()

from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

# IGNORE - Just to print AI response Beautifully
from IPython.display import Markdown
```

## Step 2: Get Image Path in Python

```python
path = Path("./media/cat-and-spidey.jpg")
path
```

## Step 3: Convert Image &rarr; **Image Bytes**

```python
image_bytes = path.read_bytes()
print(image_bytes)
```

## Step 4: Convert Image Bytes &rarr; **Base64 Bytes**

```python
base64_bytes = base64.b64encode(image_bytes)
print(base64_bytes)
print(type(base64_bytes))
```

## Step 5: Convert Base64 Bytes to &rarr; **Base64 String** (utf-8) format

But why?
- LLM can't read Byte data, also JSON cannot transfer it, they can only read strings or text
- Thus, we need to convert Byte to String

What is **UTF-8**?
- **In short**:
    - `Hello World` is for humans.
    - `01001000 01100101...` is for computers.
    - UTF-8 provides `rules` about how to convert this string to bits.
- **More Information**:
    - **8-bit Unicode Transformation Format** is a standard language representation format.
    - Characters like `A`, `B`, `C`, etc. or `$`, `&`, `*`, etc. all are represented using 8-bits, 16-bits, 32-bits, etc. (e.g., `00000000`, `11111111`, `0100100101001001`, etc.) values
    - All Web Browsers support this format
    - This format is also compatible with ASCII format

```python
base64_string = base64_bytes.decode("utf-8")
print(base64_string)
print(type(base64_string))
```

## Step 6: Provide Base64 String to LLM.

We also need to specify **MIME Type** (Multipurpose Internet Mail Extensions).

MIME Type means what is the extention or type of the media being loaded, whether it is a JPG/JPEG image, PNG Image, GIF, MP4, etc.

If MIME Type is not specified, the LLM may consider given `byte_string` as a normal string instead of a file OR may raise an error.

> NOTE: MIME type for JPEG is ALWAYS "image/jpeg", never "image/jpg".

```python
from langchain_core.messages import SystemMessage, HumanMessage

messages = [
    # System Message
    SystemMessage(content="Explain what is given in this image in 50 words."),

    # Standard LangChain Multimodal Format (Supported by Gemini & OpenAI)
    HumanMessage(
        content=[
            {
                "type": "image_url",
                "image_url": {
                    "url": f"data:image/jpeg;base64,{base64_string}"
                },
            }
        ]
    )
]

# NOTE FOR OPENAI NATIVE API (NON-LANGCHAIN):
# OpenAI's raw SDK expects:
# {"type": "image_url", "image_url": {"url": f"data:image/jpeg;base64,{base64_string}"}}
```

## Step 7: Enjoy the Output

```python
ai_message = llm.invoke(messages)
Markdown(ai_message.content)
```
